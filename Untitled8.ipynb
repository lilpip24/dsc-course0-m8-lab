```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

# Load our local CSV datasets from Google Colab
wine_df = pd.read_csv('wine.csv')
chick_df = pd.read_csv('chickwts.csv')
arrests_df = pd.read_csv('USArrests.csv')

# Check structure and clean up unnamed index columns if present
for df in [wine_df, chick_df, arrests_df]:
    unnamed_cols = [c for c in df.columns if 'Unnamed' in c]
    if len(unnamed_cols) > 0:
        df.drop(columns=unnamed_cols, inplace=True)

print("Wine shape:", wine_df.shape)
print("Chickwts shape:", chick_df.shape)
print("USArrests shape:", arrests_df.shape)

# Verify no missing values across all files
print("\nMissing values check:")
print("Wine missing:", wine_df.isnull().sum().sum())
print("Chickwts missing:", chick_df.isnull().sum().sum())
print("USArrests missing:", arrests_df.isnull().sum().sum())

# Separate Wine features and standard scale them
if 'target' in wine_df.columns:
    X_wine_raw = wine_df.drop(columns=['target'])
    y_wine = wine_df['target']
elif 'Class' in wine_df.columns:
    X_wine_raw = wine_df.drop(columns=['Class'])
    y_wine = wine_df['Class']
else:
    # If target is the last column
    X_wine_raw = wine_df.iloc[:, :-1]
    y_wine = wine_df.iloc[:, -1]

scaler_wine = StandardScaler()
X_wine_scaled = scaler_wine.fit_transform(X_wine_raw)

# Preview the first few rows of each to make sure everything looks good
print("\nWine preview:")
print(wine_df.head(2))

print("\nChickwts preview:")
print(chick_df.head(2))

print("\nUSArrests preview:")
print(arrests_df.head(2))
```

```python
from sklearn.decomposition import PCA
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import classification_report, accuracy_score

# 1. Apply PCA to retain 95% of the variance
pca_wine = PCA(n_components=0.95, random_state=42)
X_wine_pca = pca_wine.fit_transform(X_wine_scaled)

print("Original feature count:", X_wine_raw.shape[1])
print("PCA feature count (95% variance):", X_wine_pca.shape[1])

# 2. Split dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X_wine_pca, y_wine, test_size=0.3, random_state=42, stratify=y_wine
)

# 3. Setup GridSearchCV to find best k-value and distance metric
search_space = {
    'n_neighbors': list(range(1, 16)),
    'metric': ['euclidean', 'manhattan']
}

grid_knn = GridSearchCV(
    KNeighborsClassifier(),
    search_space,
    cv=5,
    scoring='accuracy'
)
grid_knn.fit(X_train, y_train)

# 4. Train model with best parameters and evaluate
best_knn = grid_knn.best_estimator_

print("\nBest k-NN Parameters:", grid_knn.best_params_)

y_pred = best_knn.predict(X_test)

print("\nAccuracy Score:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

```python
from sklearn.metrics.pairwise import cosine_similarity

# 1. Aggregate weight stats for each feed type
feed_stats = chick_df.groupby('feed')['weight'].agg(
    mean_weight='mean',
    std_weight='std',
    median_weight='median',
    min_weight='min',
    max_weight='max'
).reset_index()

# Handle any potential NaN standard deviations if groups are tiny
feed_stats.fillna(0, inplace=True)

# 2. Standardize numerical summary features
scaler_feed = StandardScaler()
feed_numeric = feed_stats.drop(columns=['feed'])
feed_scaled = scaler_feed.fit_transform(feed_numeric)

# 3. Apply PCA to reduce features to 1 principal component
pca_feed = PCA(n_components=1, random_state=42)
feed_pca = pca_feed.fit_transform(feed_scaled)

# 4. Compute cosine similarity matrix between feed types
sim_matrix = cosine_similarity(feed_pca)
sim_df = pd.DataFrame(
    sim_matrix,
    index=feed_stats['feed'],
    columns=feed_stats['feed']
)

print("Feed Similarity Matrix (Cosine Similarity):")
print(sim_df.round(3))

# Helper function to get top feed recommendations
def get_similar_feeds(target_feed, top_n=2):
    if target_feed not in sim_df.index:
        return f"'{target_feed}' not found in dataset."

    # Sort similarity scores in descending order and skip the self-match
    feed_scores = sim_df[target_feed].sort_values(ascending=False)
    filtered_scores = feed_scores.drop(target_feed)

    return filtered_scores.head(top_n)

# Example output for stakeholder demo
sample_feed = sim_df.index[0]
print(f"\nTop recommendations for '{sample_feed}':")
print(get_similar_feeds(sample_feed, top_n=2))
```

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.mixture import GaussianMixture

# 1. Feature selection and scaling
crime_cols = ['Murder', 'Assault', 'Rape']
X_crime_raw = arrests_df[crime_cols]

scaler_crime = StandardScaler()
X_crime_scaled = scaler_crime.fit_transform(X_crime_raw)

# 2. PCA reduction to 2 components
pca_crime = PCA(n_components=2, random_state=42)
X_crime_pca = pca_crime.fit_transform(X_crime_scaled)

# 3. Determine optimal clusters using Elbow Method and BIC
k_values = range(1, 7)
inertia_scores = []
bic_scores = []

for k in k_values:
    # K-Means Inertia
    km_test = KMeans(n_clusters=k, random_state=42, n_init=10)
    km_test.fit(X_crime_pca)
    inertia_scores.append(km_test.inertia_)

    # GMM BIC
    gmm_test = GaussianMixture(n_components=k, random_state=42)
    gmm_test.fit(X_crime_pca)
    bic_scores.append(gmm_test.bic(X_crime_pca))

# Plot Elbow and BIC curves
fig, axes = plt.subplots(1, 2, figsize=(13, 4))

axes[0].plot(k_values, inertia_scores, marker='o', color='b')
axes[0].set_title('K-Means Elbow Method (Inertia)')
axes[0].set_xlabel('Number of Clusters (k)')
axes[0].set_ylabel('Inertia')

axes[1].plot(k_values, bic_scores, marker='o', color='orange')
axes[1].set_title('GMM BIC Curve')
axes[1].set_xlabel('Number of Clusters (k)')
axes[1].set_ylabel('BIC Score')

plt.tight_layout()
plt.show()

# 4. Apply final models with optimal k (k=3 based on elbow/BIC)
chosen_k = 3

kmeans_final = KMeans(n_clusters=chosen_k, random_state=42, n_init=10)
kmeans_labels = kmeans_final.fit_predict(X_crime_pca)

gmm_final = GaussianMixture(n_components=chosen_k, random_state=42)
gmm_labels = gmm_final.fit_predict(X_crime_pca)

# Visualize and compare clustering assignments
plt.figure(figsize=(13, 5))

plt.subplot(1, 2, 1)
plt.scatter(X_crime_pca[:, 0], X_crime_pca[:, 1], c=kmeans_labels, cmap='viridis', s=60)
plt.title('K-Means Clustering (Hard Assignment)')
plt.xlabel('PC1')
plt.ylabel('PC2')

plt.subplot(1, 2, 2)
plt.scatter(X_crime_pca[:, 0], X_crime_pca[:, 1], c=gmm_labels, cmap='plasma', s=60)
plt.title('GMM Clustering (Probabilistic Assignment)')
plt.xlabel('PC1')
plt.ylabel('PC2')

plt.tight_layout()
plt.show()
```
